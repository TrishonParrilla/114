## Stacks

A stack is an ordered collection of elements where items are added and removed from one end, commonly refering to as the "top". 
The ordering principle is referred to as LIFO (Last in, first out)

```python
stack = []
stack.append("A")
stack.append("B")
stack.append("C")
top = stack.pop()
print(stack)
print(top)
print(stack.pop())
stack.append("D")
print(stack.pop())
print(stack)

```

```python
#simplified version(relying on built in functions)
class Stack:
    def __init__(self):
        self.items = []
    
    def push(self, elements):
        self.items.append(elements)

    def pop(self):
        return self.items.pop()
    
    def peek(self):
        return self.items[len(self.items)-1]
    
    def size(self):
        return len(self.items)

    def is_empty(self):
        return self.items == []

    def __str__(self):
        print(self.items)
```

```python
stack_obj = Stack()

stack_obj.push("A")
stack_obj.push("B")
stack_obj.push("C")

top = stack_obj.pop()
print(stack_obj.items)
print(top)
print(stack_obj.peek())
print(stack_obj.size())
```
