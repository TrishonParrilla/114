## Stacks

A stack is an ordered collection of elements where items are added and removed from one end, commonly refering to as the "top". 
The ordering principle is referred to as LIFO (Last in, first out)

```python
stack = []
stack.append("A")
stack.append("B")
stack.append("C")
top = stack.pop()
print(stack)
print(top)
print(stack.pop())
stack.append("D")
print(stack.pop())
print(stack)

```

```python
#simplified version(relying on built in functions)
class Stack:
    def __init__(self):
        self.items = []
    
    def push(self, elements):
        self.items.append(elements)

    def pop(self):
        return self.items.pop()
    
    def peek(self):
        return self.items[len(self.items)-1]
    
    def size(self):
        return len(self.items)

    def is_empty(self):
        return self.items == []

    def __str__(self):
        return f"Stack -{self.items}"
```

```python
stack_obj = Stack()

stack_obj.push("A")
stack_obj.push("B")
stack_obj.push("C")

top = stack_obj.pop()
print(stack_obj.items)
print(top)
print(stack_obj.peek())
print(stack_obj.size())
```

```python
class CustomStack():
    class Node:
        def __init__(self, data):
            self.data = data
            self.below = None
        def __str__(self):
            return f"Node ({self.data})"

    def __init__(self):
            self.top = None
            self.length = 0

    def push(self, value):
        self.length += 1
        if not self.top:
            self.top = self.Node(value)
        
        else:
            new_node = self.Node(value)
            new_node.below = self.top
            self.top = new_node
        
    
    def pop(self):

        if not self.top:
            raise IndexError("Stack is empty")
        else:
            self.length -= 1
            old_value = self.top.data
            self.top = self.top.below
            return old_value #this is to have a value to use after popping it
            
    
    def is_empty(self):
        if self.top == None:
            return True    
        else:
            return False
        
    def peek(self): #returns value at the top without removing it
        if self.is_empty():
            raise IndexError("stack is empty")
        else:
            return self.top()
    

    def size(self): # returns amount of elements in stack
        if self.is_empty():
            raise IndexError("Stack is empty")
        else:
            count = 0
            current_node = self.top
            while not current_node is None:
                count += 1
                current_node = current_node.below
            return count
    
    
    def __str__(self):
        if self.is_empty():
            raise IndexError("Stack is Empty")
        else:
            append_node = ""
            current_node = self.top
            while not current_node is None:
                if current_node.below is None:
                    append_node += f"{current_node.data}"
                else:
                    append_node += f"{current_node.data},"
                current_node = current_node.below
        return f"CustomStack - [{append_node}]"
    


    
```

```python
cstack = CustomStack()

cstack.push("A")
cstack.push("B")
cstack.push("C")
cstack.push(1)
print(cstack)
print(cstack.size())
print(cstack)


```

```python
def invert_string(string):
    stack = CustomStack()
    for char in string:
        stack.push(char)
    out = ""
    while not stack.is_empty():
        out += stack.pop()
    return out
invert_string("ih")
      
        
```
