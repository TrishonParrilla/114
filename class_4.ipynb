# Class 4 - CustomQueue and LinkedLists

```python
#Custom Queue Class built from scratch

class CustomQueue():
    class Node:
        def __init__(self, data):
            self.data = data
            self.next = None
    def __init__(self):
        self.front = None
        self.rear = None
        self.length = 0
    
    def is_empty(self):
        return self.front is None
    
    def enque(self, data):
        self.length += 1
        new_node = self.Node(data) #create a node with desired data
        if self.is_empty(): #checks if que is empty
            self.front = new_node
            self.rear = new_node # giving both pointers the value of new node to point to itself
        else:
            self.rear.next = new_node #from the existing rear node, set the next property to the new node
            self.rear = new_node  # update the reference of rear to the new node

    def deque(self):
        if self.is_empty():
            raise IndexError("Queue Is Empty") 
        else:
            self.length -= 1 #subtract from the length property
            dequed_node = self.front #store in a new variable the current front
            self.front = self.front.next #set the new front to be the next propterty from the current one
            if self.is_empty(): # if we dequed the last node
                self.rear = None #update the rear and front referenes to none
                self.front = None
            return dequed_node.data #return the data
    
    def size(self):
        return self.length
    
    def peek(self):
        if self.is_empty():
            raise IndexError("Queue is empty")
        else:
            return self.front.data
        
    def __str__(self):
        if self.is_empty():
            raise IndexError("Queue is Empty")
        else: 
            append_node = ""
            current_node = self.front
            while not current_node is None:
                if current_node == self.rear:
                    append_node += f"{current_node.data}"
                else:
                    append_node += f"{current_node.data}, "
                current_node = current_node.next
            return f"customQueue - [{append_node}]"
```

```python
obj = CustomQueue()
obj.enque("A")
obj.deque()
print(obj.is_empty())
```

# Challenge: Reverse a list using a queue
1. Given a list:
    - numbers = [1, 2, 3, 4, 5]

2. use a queue to reverse the order of numbers
3. print the reverse list

```python
numbers = CustomQueue()
reversed_num = []
numbers.enque(1)
numbers.enque(2)
numbers.enque(3)
numbers.enque(4)
numbers.enque(5)
print(numbers)
for i in range (0, (numbers.size())):
    reversed_num.insert(0, numbers.deque())
print(reversed_num)
```
