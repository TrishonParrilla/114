# Class 3 - Queue

A queue is an ordered collection of elements where items are added and removed from **opposite** ends,
commonly refered to as the "front" and "back" or "rear" of the queue. The ordering principle is referred
to as FIFO(first in, First out) or "First come first served"

```python
#simplified Queue (with built in functions)

class Queue:
    def __init__(self):
        self.items = []

    def enque(self, element):
        self.items.insert(0,element)
    
    def deque(self):
        return self.items.pop()
    
    def __str__(self):
        return f"Queue - {self.items}"
    
    #Nice to have methods
    def peek(self):
        return self.items[len(self.items)-1]
    
    def size(self):
        return len(self.items)
    
    def is_empty(self):
        return self.items == []
```

```python
my_queue = Queue()
my_queue.enque("A")
my_queue.enque("B")
my_queue.enque("C")
print(my_queue)
value = my_queue.deque()
print(value)
my_queue.enque("Z")
print(my_queue)
```
