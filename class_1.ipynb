## Psuedocode Challenge 2: Given a list of numbers, determine which one is largest.

START
    INITIALIZE list of numbers = [1, 9, 15, 3, 6, 10]
    
    For Number in Numbers
        
        IF number is > final index in Numbers
            
            Append Number
    
    Return final index of number 
END

START 

Create a variable for the Largest

Set largest to first value in the List

For each number in the list starting at the SECOND item

if the number is greater than the largest

update largest to this number

display largest

END




```python
nums = [12,4,27, 3, 1000, 50, 7]

largest = nums[0]

for i in range(1, len(nums)): #generates a range as long as  1 to the length of the list 
    if nums[i] > largest: # a number in that range then is matched to a specific index of the list  
        largest = nums[i]  #largest is now given the value of that number that was matched 
nums[0] = largest

print(largest)       
print(nums) 

#Technically we dont HAVE to iterate the range of indexes. We only do this because we want to change the first index to reflect as part of our solution
# we can also just iterate through the values of the list directly and compare them to our initialized variable for biggest num and reassign our biggest variable to store the value of our iterable
```

## Total numbers from the collection that are even. 
The collection is numbers = [3,5,27,20,6,3.5,60]
    Is it the number of how many are even?
    Is it the sum of all the even numbers?

Start
    Initialize the list[nums]
    Initialize a variable named counter with value of 0
    For every num in the list
        IF the number is divisible by 2 with a remainder of 0
            INCREMENT VAL by 1
    Display VAL
End

```python
nums = [3,5,27,20,6,3.5,60]
counter = 0
for num in nums: #dont need the index so we can just directly iterate through the values of the list rather than their index
    if num % 2 == 0:
        counter+= 1
print(counter)
```
